```python
import pandas as pd
import numpy as np



df=pd.read_csv('/content/drive/MyDrive/TransformersCode/e_commerce/ProductRecommendations/flipkart_com-ecommerce_sample.csv')


embeddings=np.load('/content/drive/MyDrive/TransformersCode/e_commerce/ProductRecommendations/product_embeddings.npy')


```

```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

def get_recommendations(product_name_query, top_n=5):
  matched_rows = df[df["product_name"].str.contains(product_name_query, case=False, na=False, regex=False)]
  if matched_rows.empty:
    return f"Product '{product_name_query}' not found!"

  idx = matched_rows.index[0]
  target_name = df.loc[idx, "product_name"]

  target_embedding = embeddings[idx].reshape(1, -1)
  similarity_scores = cosine_similarity(target_embedding, embeddings)[0]

  sorted_indices = similarity_scores.argsort()[::-1]

  seen_names = {target_name}
  unique_indices = []
  for i in sorted_indices:
    name = df.loc[i, "product_name"]
    if name not in seen_names:
      seen_names.add(name)
      unique_indices.append(i)
      if len(unique_indices) == top_n:
        break

  results = pd.DataFrame({
      "product_name": df.loc[unique_indices, "product_name"],
      "similarity_score": similarity_scores[unique_indices]
  })
  return results

sample_product = "Alisha Solid Women's Cycling Shorts"
recommendations = get_recommendations(sample_product, top_n=5)
display(recommendations)
```
