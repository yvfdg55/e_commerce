```python
import pandas as pd
import zipfile

path_file='/content/drive/MyDrive/TransformersCode/e_commerce/ProductRecommendations'
zip='/content/drive/MyDrive/TransformersCode/e_commerce/ProductRecommendations/flipkart_com-ecommerce_sample.zip'
with zipfile.ZipFile(zip, 'r') as zip_ref:
    zip_ref.extractall(path_file)

df=pd.read_csv('/content/drive/MyDrive/TransformersCode/e_commerce/ProductRecommendations/flipkart_com-ecommerce_sample.csv')
df
```

```python
import re





def create_text_representation(row):
  # استخدام القيم مباشرة والتأكد من تحويلها لنص بطريقة آمنة
  name = (
      ""
      if pd.isna(row["product_name"])
      else __builtins__.str(row["product_name"])
  )
  category = (
      ""
      if pd.isna(row["product_category_tree"])
      else __builtins__.str(row["product_category_tree"])
  )
  brand = "" if pd.isna(row["brand"]) else __builtins__.str(row["brand"])
  desc = "" if pd.isna(row["description"]) else __builtins__.str(row["description"])

  # استخراج المواصفات
  spec_val = row["product_specifications"]
  spec_str = "" if pd.isna(spec_val) else __builtins__.str(spec_val)

  keys = re.findall(r'"key"\s*=>\s*"([^"]*)"', spec_str)
  values = re.findall(r'"value"\s*=>\s*"([^"]*)"', spec_str)

  # استخدام __builtins__.zip لتجنب تداخل أسماء المتغيرات إذا تم تعريف متغير باسم zip سابقاً
  specs_list = [f"{k}: {v}" for k, v in __builtins__.zip(keys, values)]
  specs_text = " | ".join(specs_list)

  return f"Product: {name} | Category: {category} | Brand: {brand} | Specs: {specs_text} | Description: {desc}"


# تطبيق الدالة على الجدول
df["combined_text"] = df.apply(create_text_representation, axis=1)
print("تم دمج النصوص بنجاح!")
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

product_embeddings = model.encode(
    df["combined_text"].tolist(), show_progress_bar=True
)
```

```python
import numpy as np

np.save("/content/drive/MyDrive/TransformersCode/e_commerce/ProductRecommendations/product_embeddings.npy", product_embeddings)
df.to_pickle("processed_data.pkl")

print("Finished successfully! Embeddings and processed data have been saved.")
```
